```python
#from pyspark.sql import SparkSession
from pyspark.sql import functions as F

#spark = SparkSession.builder.getOrCreate()

# Sample data
data = [("Alice", 25, "HR"), ("Bob", 30, "IT"), ("Charlie", 35, "Finance")]
columns = ["name", "age", "department"]
df = spark.createDataFrame(data, columns)

# 1. Select specific columns using strings (SQL: SELECT name, department FROM table)
df.select("name", "department").show()

# 2. Select using F.col() to transform or calculate columns
# SQL: SELECT name, age + 5 AS age_in_five_years FROM table
df.select(
    F.col("name"),
    (F.col("age") + 5).alias("age_in_five_years")
).show()

# 3. Mixing SQL expressions directly inside selectExpr()
# SQL: SELECT name, age * 2 AS double_age FROM table
df.selectExpr("name", "age * 2 AS double_age").show()
```

```python
# New

from pyspark.sql import Row

df2 = spark.createDataFrame([
    Row(age=10, height=80.0, NAME="Alice"),
    Row(age=10, height=80.0, NAME="Alice"),
    Row(age=5, height=float("nan"), NAME="BOB"),
    Row(age=None, height=None, NAME="Tom"),
    Row(age=None, height=float("nan"), NAME=None),
    Row(age=9, height=78.9, NAME="josh"),
    Row(age=18, height=1802.3, NAME="bush"),
    Row(age=7, height=75.3, NAME="jerry"),
])

df2.show()
```

```python
from pyspark.sql import functions as F

df2.select(
    F.col("age").isNull()
).show()
```

```python
from pyspark.sql import functions as F

df2.where(F.col("age").isNull()).select(
    "NAME", 
    "age",
    "height"
).show()
```

```python
from pyspark.sql import functions as F

df2.where(F.col("age").isNull()).select(
    "*"
).show()
```

```python
total_rows = df2.count()
print(f"Total rows: {total_rows}")
# Output: Total rows: 8
```

```python
# NOT RECOMMENDED for large datasets
for c in df2.columns:
    c_nulls = df2.where(F.col(c).isNull()).count()
    print(f"{c}: {c_nulls}")
```

```python
from pyspark.sql import functions as F

null_counts_df = df2.select(
    [
        F.count(F.when(F.col(c).isNull(),1)).alias(c)
        for c in df2.columns
    ]
)

null_counts_df.show()
```
